(2.5pkt) Uzupełnij klasę liczb zespolonych omawianą na zajęciach o następujące operacje
* porównywanie liczb zespolonych (operatory `==` i `!=`)
* odejmowanie liczby całkowitej/rzeczywistej od liczby zespolonej
* odejmowanie liczby zespolonej od liczby całkowitej/rzeczywistej
* negacja liczby zespolonej
* mnożenie liczb zespolonych
* mnożenie liczby zespolonej i liczby całkowitej/rzeczywistej (lewo- i prawostronnie)
* dzielenie liczb zespolonych (jeżeli dzielimy przez 0 powinno zostać zwróceone `None`)
* dzielenie liczby zespolonej przez liczbę całkowitą/rzeczywistą
* dzielenie liczby całkowitej/rzeczywistej przez liczbę zespoloną
* obliczanie modułu liczby zespolonej
* obliczanie argumentu liczby zespolonej
* wypisywanie liczby zespolonej w postaci wykładniczej, np.: liczbę $3+j4$ należy wypisać jako `5exp(j0.9272)`

W większości przypadków należy przeładować odpowiednie operatory.
Wyniki operacji umieszczamy w nowych obiektach klasy `Cmplx`.

Nastepnie należy przetestować i przedstawić działanie klasy i jej metod
dla liczb zespolonych o współczynnikach wylosowanych z przedziału $(-10,10)$.


```python
import math


class Cmplx:

    def __init__(self, re, im):
        self._im = im
        self._re = re

    def __str__(self):
        if self._im>=0:
            return str(self._re)+"+j"+str(self._im)
        else:
            return str(self._re)+"-j"+str(-self._im)

    @property
    def re(self):
        return self._re

    @property
    def im(self):
        return self._im
    
    def __eq__(self, other):
        return self.re == other.re and self.im == other.im
    
    def __ne__(self, other):
        return not self.__eq__(other)
            
    def __sub__(self, other):
        if type(other) == Cmplx:
            return Cmplx(self.re-other.re, self.im-other.im)
        else:
            return Cmplx(self.re-other, self.im)
    
    def __rsub__(self, other):
        return Cmplx(other - self.re, -self.im)
    
    def __neg__(self):
        return Cmplx(-self._re, -self.im)

    def __mul__(self, other):
        if type(other) == Cmplx:
            return Cmplx(self.re * other.re - self.im * other.im, self.re * other.im + self.im * other.re)
        else:
            return Cmplx(self.re * other, self.im * other)
    
    def __rmul__(self, other):
        return self.__mul__(other)
    
    def __truediv__(self, other):
        if type(other) == Cmplx:
            if other.re == 0 and other.im == 0:
                return None
            else:
                denom = other.re ** 2 + other.im ** 2
                re = (self.re * other.re + self.im * other.im) / denom
                im = (self.im * other.re - self.re * other.im) / denom
                return Cmplx(re, im)
        else:
            if other == 0:
                return None
            else:
                return Cmplx(self.re / other, self.im / other)
    
    def __rtruediv__(self, other):
        if self.re == 0 and other.im == 0:
            return None
        else:
            denom = self.re ** 2 + self.im ** 2
            re = (other * self.re) / denom
            im = (-other * self.im) / denom
            return Cmplx(re, im)
        
    def __abs__(self):
        return math.sqrt(self.re ** 2 + self.im ** 2)
    
    def argument(self):
        if self.re == 0:
            if self.im > 0:
                return math.pi / 2
            elif self.im < 0:
                return -math.pi / 2
            else:
                return 0
        else:
            argument = math.atan(abs(self.im) / abs(self.re))
            if self.re > 0 and self.im >= 0:
                return argument
            elif self.re < 0 <= self.im:
                return math.pi - argument
            elif self.re < 0 and self.im < 0:
                return -(math.pi - argument)
            else:
                return -argument

    def exp_form(self):
        modulus = abs(self)
        argument = self.argument()
        return f"{modulus}exp(j{argument})"

import random

a = Cmplx(random.randint(-10, 10), random.randint(-10, 10))
b = Cmplx(random.randint(-10, 10), random.randint(-10, 10))

print("Liczba a:", a)
print("Liczba b:", b)

print("Czy a i b są równe?", a == b)
print("Czy a i b są różne?", a != b)

print("a - 1:", a - 1)
print("2 - b:", 2 - b)

print("Negacja liczby -a:", -a)

print("Iloczyn a * b:", a * b)
print("a * 2:", a * 2)
print("3 * b:", 3 * b)

print("Dzielenie liczby a / b:", a / b)
print("a / 2:", a / 2)
print("4 / b:", 4 / b)

print("Moduł liczby a:", abs(a))
print("Moduł liczby b:", abs(b))

print("Argument liczby a:", a.argument())
print("Argument liczby b:", b.argument())

print("Liczba a w postaci wykładniczej:", a.exp_form())
print("Liczba b w postaci wykładniczej:", b.exp_form())
```

(2.5pkt) Zaimplementuj klasę `Poly` reprezentującą wielomian.
Klasa powinna przechowywać współczynniki wielomianu i umożliwiać nastepujące operacje
* tworzenie nowego wielomianu na podstawie listy współczynników
* wypisywanie wielomianu w ładnej formie np: `2x^3+5x^2-3`
* porównywanie wielomianów (operatory `==` i `!=`)
* operacje arytmetyczne na wielomianach: dodawanie, odejmowanie, mnożenie, negacja
* mnożenie wielomianu przez liczbę (lewo- i prawostonne)
* obliczanie stopnia wielomianu

Dodatkowo zaimplementować funkcjonalność obliczania wartości wielomianu dla konktretnej liczby.
Wykorzystać do tego metodę specjalną `__call__()`, która umożliwia "wywoływanie obiektu" tak jak gdyby był funkcją np.:

`a = Poly([1,2,3])`
`a(5)`

W większości przypadków należy przeładować odpowiednie operatory.
Wyniki operacji umieszczamy w nowych obiektach klasy `Poly`

Nastepnie należy przetestować i przedstawić działanie klasy i jej metod
dla wielomianów o stopniach z przedziału $(0,10)$ i współczynnikach wylosowanych z przedziału $(-10,10)$.


```python
class Poly:
    def __init__(self, coefficients):
        self.coefficients = coefficients

    def __str__(self):
        result = ""
        degree = len(self.coefficients) - 1
        for coef in self.coefficients:
            if coef != 0:
                if degree > 1:
                    result += f"{coef}x^{degree} + "
                elif degree == 1:
                    result += f"{coef}x + "
                else:
                    result += f"{coef}"
            degree -= 1
        return result.rstrip(" + ")

    def __eq__(self, other):
        return self.coefficients == other.coefficients

    def __ne__(self, other):
        return not self.__eq__(other)

    def __add__(self, other):
        max_length = max(len(self.coefficients), len(other.coefficients))
        result = [0] * max_length
        for i in range(len(self.coefficients)):
            result[i] += self.coefficients[i]
        for i in range(len(other.coefficients)):
            result[i] += other.coefficients[i]
        return Poly(result)

    def __sub__(self, other):
        max_length = max(len(self.coefficients), len(other.coefficients))
        result = [0] * max_length
        for i in range(len(self.coefficients)):
            result[i] += self.coefficients[i]
        for i in range(len(other.coefficients)):
            result[i] -= other.coefficients[i]
        return Poly(result)

    def __neg__(self):
        return Poly([-coef for coef in self.coefficients])

    def __mul__(self, other):
        if type(other) == Poly:
            result = [0] * (len(self.coefficients) + len(other.coefficients) - 1)
            for i in range(len(self.coefficients)):
                for j in range(len(other.coefficients)):
                    result[i + j] += self.coefficients[i] * other.coefficients[j]
            return Poly(result)
        else: 
            return Poly([coef * other for coef in self.coefficients])
        

    def __rmul__(self, other):
        return Poly([other * coef for coef in self.coefficients])

    def degree(self):
        return len(self.coefficients) - 1

    def __call__(self, x):
        result = 0
        max_degree = self.degree()
        for i in range(len(self.coefficients)):
            result += self.coefficients[i] * (x ** max_degree)
            max_degree -= 1
        return result

import random

coefficients1 = [random.randint(-10, 10) for _ in range(random.randint(1, 10))]
coefficients2 = [random.randint(-10, 10) for _ in range(random.randint(1, 10))]
poly1 = Poly(coefficients1)
poly2 = Poly(coefficients2)

print("Wielomian poly1:", poly1)
print("Wielomian poly2:", poly2)

print("Czy wielomiany poly1 i poly2 są równe?", poly1 == poly2)
print("Czy wielomiany poly1 i poly2 są różne?", poly1 != poly2)

print("Suma wielomianów poly1 + poly2:", poly1 + poly2)
print("Różnica wielomianów poly1 - poly2:", poly1 - poly2)
print("Negacja wielomianu -poly1:", -poly1)
print("Iloczyn wielomianów poly1 * poly2:", poly1 * poly2)
print("Iloczyn wielomianu poly1 * 2:", poly1 * 2)
print("Iloczyn wielomianu 2 * poly1:",  2 * poly1)
print("Stopień wielomianu poly1:", poly1.degree())
print("Stopień wielomianu poly2:", poly2.degree())

x = random.randint(-10, 10)
print(f"Wartość wielomianu poly1 dla x={x}:", poly1(x))
print(f"Wartość wielomianu poly2 dla x={x}:", poly2(x))
```
