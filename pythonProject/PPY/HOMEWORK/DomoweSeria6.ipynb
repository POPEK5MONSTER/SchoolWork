Napisać generator, który generuje kolejne dzielniki liczby naturalnej $N$.
Wypisać dzielniki wybranej liczby $N$ korzystając z pętli for i generatora.
Utworzyć listę dzielników wybranej liczby $N$ korzystając z list comprehensions.

````python
def dzielniki(N):
    for i in range(1, N+1):
        if N % i == 0:
            yield i

def lista_dzielnikow(N):
    return [dzielnik for dzielnik in dzielniki(N)]

wybrana_liczba = 36
print("Lista dzielników:", lista_dzielnikow(wybrana_liczba))
````

Napisać generator, który generuje pary (dzielnik pierwszy, krotność dzielnika) dla liczby $N$.
Wypisać dzielniki pierwsze wybranej liczby $N$ wraz z krotnościami korzystając z pętli for i generatora.
Utworzyć słownik o kluczach będących dzielnikami pierwszymi wybranej liczby $N$
i wartościach będących krotnościami odpowiednich dzielników korzystając z dictionary comprehensions.

````python
def generuj_pary_dzielnikow(N):
    dzielnik = 2
    while N > 1:
        krotnosc = 0
        while N % dzielnik == 0:
            krotnosc += 1
            N //= dzielnik
        if krotnosc > 0:
            yield dzielnik, krotnosc
        dzielnik += 1

def slownik_dzielnikow_z_krotnosciami(N):
    return {dzielnik: krotnosc for dzielnik, krotnosc in generuj_pary_dzielnikow(N)}

wybrana_liczba = 100
print("Słownik dzielników z krotnościami:", slownik_dzielnikow_z_krotnosciami(wybrana_liczba))
````

Napisać generator, który dla danej listy liczb, generuje wielomiany o współczynnikach będących liczbami
z listy i coraz niższych potęgach. Na przykład dla listy liczb `[3,5,0,6,4]` generator ma generować kolejno wielomiany

$$ 3x^4,\ 5x^3,\ 0,\ 6x,\ 4 $$

Wykorzystać wbudowaną funkcję sum() aby zsumować wielomiany generowane za pomocą generatora
dla przykładowej listy listy liczb. Porównać (operator `==`) wielomian uzyskany tą metodą
z wielomianem utworzonym bezpośrednio na podstawie listy liczb.


````python
class Poly:
    def __init__(self, coefficients):
        self.coefficients = coefficients

    def __repr__(self):
        if not self.coefficients or self.coefficients == [0 for _ in range(len(self.coefficients))]:
            return "0"
        result = ""
        degree = len(self.coefficients) - 1
        for coef in self.coefficients:
            if coef != 0:
                if degree > 1:
                    result += f"{coef}x^{degree} + "
                elif degree == 1:
                    result += f"{coef}x + "
                else:
                    result += f"{coef}"
            degree -= 1
        return result.rstrip(" + ")
    
    def __add__(self, other):
        if type(self) == type(other):
            if len(self.coefficients) > len(other.coefficients):
                result = self.coefficients
                for i in range(-1, -len(other.coefficients) - 1, -1):
                    result[i] += other.coefficients[i]
                return Poly(result)
            else:
                result = other.coefficients
                for i in range(-1, -len(self.coefficients) - 1, -1):
                    result[i] += self.coefficients[i]
                return Poly(result)
        elif type(other) == type(1):
            result = self.coefficients
            result[-1] += other
            return Poly(result)
        else:
            return None
        
    def __radd__(self, other):
        return self.__add__(other)
    
    def __eq__(self, other):
        return self.coefficients == other.coefficients

def poly_generator(coefficients):
    j = 0
    while j < len(coefficients):
        yield Poly([coefficients[j]] + [0] * (len(coefficients) - j - 1))
        j += 1        
        
l = [3, 5, 0, 6, 4]

l1 = [x for x in poly_generator(l)]
wielomian2 = Poly(l)


print(l1)
suma = sum(i for i in poly_generator(l))
print(suma)
print(wielomian2)

print(suma == wielomian2)
````

Napisz dekorator  `mulArgs`, który dekoruje funkcje matematyczne jednej zmiennej.
Po udekorowaniu funkcja powinna przyjmować dowolną (zmienną) liczbę argumentów i zwracać listę wartości
dekorowanej funkcji dla tych argumentów. Na przykład funkcja

```
@mulArgs
def myExp(x):
    return exp(x)
```
powinna umożliwiać wywołanie `myExp(1,2,3,4)` i zwrócić listę
$$
\left[ e^1, e^2, e^3, e^4 \right]
$$

Udekorować funkcje `mySin, myCos, myExp` i przetestować ich działanie.


````python
import math

def mulArgs(func):
    def wrapper(*args):
        return [func(arg) for arg in args]
    return wrapper

@mulArgs
def myExp(x):
    return math.exp(x)

@mulArgs
def mySin(x):
    return math.sin(x)

@mulArgs
def myCos(x):
    return math.cos(x)

print("Eksponentywa:", myExp(1, 2, 3, 4))
print("Sinus:", mySin(1, 2, 3, 4))
print("Cosinus:", myCos(1, 2, 3, 4))
````
